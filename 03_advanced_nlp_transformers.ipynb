```python
"""
📊 SECTION 1: SENTIMENT ANALYSIS USING TEXTBLOB
This section demonstrates basic lexicon-based sentiment analysis utilizing the TextBlob library.
We measure polarity (ranging from -1.0 to 1.0) and subjectivity (ranging from 0.0 to 1.0).
"""

from textblob import TextBlob

# Source texts for sentiment analysis
var1 = "You know, ChatGPT generated a poem about remembering someone who was lost somewhere in the past."
var2 = "After some refining, I realized that I can write poetry too; if you know what I mean."

# Initialize TextBlob objects
b1 = TextBlob(var1)
b2 = TextBlob(var2)

print(f"First Sentence's sentiment is: Polarity = {b1.sentiment.polarity:.2f}, Subjectivity = {b1.sentiment.subjectivity:.2f}")
print(f"Second Sentence's sentiment is: Polarity = {b2.sentiment.polarity:.2f}, Subjectivity = {b2.sentiment.subjectivity:.2f}")
```

### 🧠 **Polarity**

**Definition:**
Polarity measures how **positive or negative** a text is in sentiment analysis.

**Range:**

* **–1.0** → Very negative
* **0.0** → Neutral
* **+1.0** → Very positive

---

### 💬 **Subjectivity**

**Definition:**
Subjectivity measures how **personal, emotional, or opinion-based** a statement is.

**Range:**

* **0.0** → Very objective (based on facts)
* **1.0** → Very subjective (based on feelings or opinions)


```python
"""
⚡ SECTION 2: COMPARATIVE SENTIMENT ANALYSIS WITH VADER AND FLAIR
In this section, we compare VADER (a fast, rule-based lexicon optimized for social media)
with Flair (a character-level LSTM neural network model that captures contextual semantics).
"""

import nltk
from nltk.sentiment.vader import SentimentIntensityAnalyzer
from flair.models import TextClassifier
from flair.data import Sentence

nltk.download('vader_lexicon', quiet=True)

var1 = "You know, ChatGPT generated a poem about remembering someone who was lost somewhere in the past."
var2 = "After some refining, I realized that I can write poetry too; if you know what I mean."
texts = [var1, var2]

# -- VADER Engine Analysis --
print("=== VADER Sentiment Engine ===")
vader = SentimentIntensityAnalyzer()
for text in texts:
    score = vader.polarity_scores(text)
    print(f"Text: '{text}'")
    print(f"→ Compound: {score['compound']}, Pos: {score['pos']}, Neu: {score['neu']}, Neg: {score['neg']}\n")

# -- Flair Engine Analysis --
print("=== Flair Neural Sentiment Engine ===")
flair_classifier = TextClassifier.load('sentiment')
for text in texts:
    sentence = Sentence(text)
    flair_classifier.predict(sentence)
    print(f"Text: '{text}'")
    print(f"→ Sentiment: {sentence.labels[0].value} (Confidence: {sentence.labels[0].score:.4f})\n")
```

```python
"""
🤖 SECTION 3: SENTIMENT ANALYSIS VIA TRANSFORMERS PIPELINES
This section utilizes pre-trained state-of-the-art deep learning transformers models
available on the Hugging Face Hub, optimized for diverse domains (general, social media, multi-lingual, and finance).
"""

from transformers import pipeline

# Registering models for various domains
models = {
    "DistilBERT (English)": "distilbert-base-uncased-finetuned-sst-2-english",
    "RoBERTa (Twitter)": "cardiffnlp/twitter-roberta-base-sentiment-latest",
    "Multilingual BERT": "nlptown/bert-base-multilingual-uncased-sentiment",
    "FinBERT (Finance)": "yiyanghkust/finbert-tone"
}

analyzers = {name: pipeline("sentiment-analysis", model=model_path) for name, model_path in models.items()}

var1 = "You know, ChatGPT generated a poem about remembering someone who was lost somewhere in the past."
var2 = "After some refining, I realized that I can write poetry too; if you know what I mean."
texts = [var1, var2]

for text in texts:
    print(f"\n📝 Text: {text}\n{'='*70}")
    for name, analyzer in analyzers.items():
        result = analyzer(text)[0]
        print(f"{name:25} → Label: {result['label']:12} | Confidence: {result['score']:.4%}")
```
